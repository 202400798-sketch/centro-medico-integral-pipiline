```python
%sql

CREATE SCHEMA IF NOT EXISTS bronze;
```

```python
tablas = [
    "cita",
    "diagnostico",
    "dispensacion",
    "especialidad",
    "hospitalizacion",
    "medicamento",
    "medico",
    "paciente",
    "receta",
    "detalle_receta",
    "hospitalizacion_diagnostico"
]
```

```python
for tabla in tablas:

    df = spark.read.jdbc(
        url=jdbc_url,
        table=f"hospital.{tabla}",
        properties=connection_properties
    )

    df.write \
        .mode("overwrite") \
        .saveAsTable(f"bronze.{tabla}")

    print(f"✅ {tabla} cargada correctamente")
```
