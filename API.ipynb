```python
import requests
import pandas as pd

municipios = {
    "San Salvador": (13.6929, -89.2182),
    "Mejicanos": (13.7403, -89.2131),
    "Santa Tecla": (13.6769, -89.2797),
    "Antiguo Cuscatlán": (13.6649, -89.2532),
    "Soyapango": (13.7102, -89.1399),
    "San Marcos": (13.6586, -89.1833)
}

registros = []

for municipio, coordenadas in municipios.items():

    latitud = coordenadas[0]
    longitud = coordenadas[1]

    url = "https://air-quality-api.open-meteo.com/v1/air-quality"

    params = {
        "latitude": latitud,
        "longitude": longitud,
        "hourly": "pm10,pm2_5,us_aqi",
        "forecast_days": 1,
        "timezone": "America/El_Salvador"
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:

        data = response.json()

        for i in range(len(data["hourly"]["time"])):

            registros.append({
                "fecha_hora": data["hourly"]["time"][i],
                "municipio": municipio,
                "latitud": latitud,
                "longitud": longitud,
                "pm10": data["hourly"]["pm10"][i],
                "pm2_5": data["hourly"]["pm2_5"][i],
                "us_aqi": data["hourly"]["us_aqi"][i]
            })

    else:
        print("Error en", municipio, response.status_code)

df_air = pd.DataFrame(registros)

print("Municipios procesados:", df_air["municipio"].nunique())
print("Registros obtenidos:", len(df_air))

display(df_air)
```
